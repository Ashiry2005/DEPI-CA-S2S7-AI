```python
bool 
def is_prime(num : int)->bool:
    for i in range(2, num):
        if num%i ==0:
            return False
    return True
```

```python
def factorial(num : int)->int:
    if num == 0:
        return 1
    return num * factorial(num - 1)
```

```python
factorial(5)
```

```python
def comm_div(num1 : int, num2: int) -> list[int]:
    limit = min(num1, num2)
    divisors = []
    for i in range(1, limit + 1):
        if num1 % i == 0 and num2 % i ==0:
            divisors.append(i)
    return divisors                
```

```python
comm_div(10,20)
```

```python
def reverse_word(text:str)-> list:
    lst = text.split()
    new_list = lst[::-1]
    return new_list
```

```python
def generate_pass(length:int)->str:
    characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ1234567890!@#$%^&*"
    password = ""

    for _ in range(length):
        password += random.choice(characters)
    return password    
```

```python
import string
import secrets
def generate_pass(length:int)->str:
    if length <=0:
        raise ValueError("password length must be greater than 0")
    characters = (string.ascii_letters + string.digits + string.punctuation)
    return "".join(secrets.choice(characters)for _ in range(length))
```

```python
def counter(words : str)->dict:
    unique_words = set(words)
    for word in unique_words:
        return {word: words.count(word) for word in set(words)}
```

```python
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}

def get_responses(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
        return random.choice(responses["default"])
def main():
    print("chatbot: hi how can i help u")
    while True:
        user_input = input("user: ").lower()
        response = get_responses(user_input)
        print("chatbot: ", response)

        if user_input == "goodbye":
            break
main()        
```
