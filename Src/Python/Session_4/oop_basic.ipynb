```python
class Student:

    def __init__(self, name = "test", age = 5) -> None:
        self.name = name
        self.age = age
        self.grade  = 0

    def set_grade(self, grade):
        self.grade = grade    

    def std_info(self):
        return f"hello {self.name} Student, your age is {self.age} and your grade is {self.grade}"
```

```python
s1 = Student("Ahmed", 30)
s1.set_grade(90)
s1.std_info()
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):

    def move(self):
        print("I can fly")

class Cat(Animal):

    def move(self):
        print("I can walk")       



a = Bird()
a.move()
 
```

```python
class Shape:
    def calcular_area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calcular_area(self):
        return 3.14 * self.radius ** 2

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def calcular_area(self):
        return self.side ** 2    

x = Circle(5)
y = Square(4)    
print("Area : ", x.calcular_area())
print("Area : ", y.calcular_area())
```

```python
class Model:
    def fit(self,x,y):
        raise NotImplementedError("Subclasses must implement this method")
    def predict(self,x):
        raise NotImplementedError("Subclasses must implement this method")

class LinearRegression(Model):  
    def fit(self,x,y):
        print("Fitting Linear Regression Model")
    def predict(self,x):
        print("Predicting using Linear Regression Model")


class DecisionTree(Model):
    def fit(self,x,y):
        print("Fitting Decision Tree Model")
    def predict(self,x):
        print("Predicting using Decision Tree Model")


models = [LinearRegression(), DecisionTree()]
for model in models:
    model.fit(None, None)
    model.predict(None)
```
