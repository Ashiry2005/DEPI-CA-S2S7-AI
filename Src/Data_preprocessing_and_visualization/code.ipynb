```python
def array_reshape(arr:list, rows:int = -1, cols:int = -1)->np.ndarray:
    try:
        arr = np.array(arr)
        return arr.reshape(rows, cols)
    except (TypeError, ValueError) as e:
        raise ValueError(f"invalid reshape operation: {e}")    
```

```python
vec = [1, 2, 3, 4, 5, 6]
mat = array_reshape(vec, 2, 3)
print(mat)
```
